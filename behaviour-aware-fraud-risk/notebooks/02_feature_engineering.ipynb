# 02 — Feature engineering (Person 2)

Build **customer-relative** baselines, then deviation features.
Keep column names in sync with `src/config.py` → `FEATURE_COLS`.

For each customer, using only **past** transactions:
- average amount, std amount
- typical hours
- known merchants / beneficiaries
- txn rate

Then: `amount_ratio`, `amount_zscore`, `is_unusual_hour`, `is_new_merchant`, `is_new_beneficiary`, `transaction_velocity`, `days_since_last_transaction`.
